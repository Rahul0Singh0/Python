## Wordnet Lemmatizer
Lemmitization technique is like stemming. The output we will get after lemmatization is called 'lemma', which is a root word that means the same thing. 

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus.
This class uses morphy() function to the WordNet CorpusReader class to find a lemma.

## Use Cases
1. Q&A
2. Chatbots
3. Text summarization

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS tags: 
Noun-n
verb-v
adjective-a
adverb-r
'''
print(lemmatizer.lemmatize("going"))
print(lemmatizer.lemmatize("going", pos="v"))
print(lemmatizer.lemmatize("going", pos="a"))
print(lemmatizer.lemmatize("going", pos="r"))
print(lemmatizer.lemmatize("going", pos="n"))
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]

for word in words:
    print(word + "----->" + lemmatizer.lemmatize(word, pos='n'))
```
